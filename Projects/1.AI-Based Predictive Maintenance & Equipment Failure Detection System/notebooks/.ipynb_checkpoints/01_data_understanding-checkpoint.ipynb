# 01 — Data Understanding

## Objective
Understand the dataset before applying any machine learning algorithm.

### Questions answered
1. What does each column represent?
2. What is the shape of the dataset?
3. Are there missing or duplicate values?
4. What is the class distribution?
5. What are the numerical ranges?
6. Which variables are categorical?

```python
from pathlib import Path
import pandas as pd
import numpy as np

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
DATA_PATH = PROJECT_ROOT / "data" / "raw" / "dataset.csv"

df = pd.read_csv(DATA_PATH)
print(f"Rows: {df.shape[0]:,}")
print(f"Columns: {df.shape[1]}")
display(df.head())
```

```python
# Data types and memory usage
df.info()
```

```python
# Descriptive statistics
display(df.describe(include="all").T)
```

```python
# Missing values and duplicates
summary = pd.DataFrame({
    "Missing_Count": df.isna().sum(),
    "Missing_Percentage": (df.isna().mean() * 100).round(2),
    "Unique_Values": df.nunique()
})
display(summary)

print("Duplicate rows:", df.duplicated().sum())
```

```python
# Target distribution
target_counts = df["Failure"].value_counts().sort_index()
target_pct = (df["Failure"].value_counts(normalize=True).sort_index() * 100).round(2)

target_summary = pd.DataFrame({
    "Count": target_counts,
    "Percentage": target_pct
})
display(target_summary)
```

```python
# Separate numerical and categorical columns
target = "Failure"
categorical_cols = df.select_dtypes(include=["object", "category"]).columns.tolist()
numerical_cols = df.select_dtypes(include=np.number).columns.drop(target).tolist()

print("Categorical columns:", categorical_cols)
print("Numerical columns:", numerical_cols)
```
